```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from datetime import datetime

file_name = "drumming_log.csv"

# 1. Collect your practice stats via terminal prompts
print("🥁 The Drumming Log 🥁\n")
rudiment = input("Which rudiment did you practice? (e.g., Paradiddle): ").strip().title()
bpm = int(input(f"What was your max BPM for the {rudiment}?: "))
duration = int(input("How many minutes did you practice it?: "))

# Auto-grab today's date
date_today = datetime.now().strftime("%Y-%m-%d")

# 2. Package the new entry into a pandas DataFrame
new_entry = pd.DataFrame({
    "Date": [date_today], 
    "Rudiment": [rudiment], 
    "BPM": [bpm], 
    "Duration (mins)": [duration]
})

# 3. Load existing data if the file exists, or create a new dataset
if os.path.exists(file_name):
    df = pd.read_csv(file_name)
    # Combine old data with the new entry
    df = pd.concat([df, new_entry], ignore_index=True)
else:
    df = new_entry

# Save the updated data back to the CSV
df.to_csv(file_name, index=False)
print(f"\n✅ Successfully logged {duration} minutes of {rudiment} at {bpm} BPM!")

# 4. Filter the data to only show the specific rudiment you just practiced
rudiment_history = df[df['Rudiment'] == rudiment]

# 5. Build and display the Matplotlib graph
plt.figure(figsize=(8, 5))
plt.plot(rudiment_history['Date'], rudiment_history['BPM'], marker='o', linestyle='-', color='indigo', linewidth=2)

plt.title(f"{rudiment} Speed Progress", fontsize=14, fontweight='bold')
plt.xlabel("Practice Date", fontsize=12)
plt.ylabel("Speed (BPM)", fontsize=12)

# Add a grid for readability and rotate dates so they don't overlap
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()

# Render the graph inside your notebook
plt.show()
```
